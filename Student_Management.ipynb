```python
bins = [17, 20, 23, df['age'].max()]
labels = ['18-20', '21-23', '24+']
df['age_group'] = pd.cut(df['age'], bins=bins, labels=labels)
age_counts = df['age_group'].value_counts()

plt.figure(figsize=(8, 6))
plt.pie(
    age_counts, 
    labels=age_counts.index, 
    autopct='%1.1f%%', 
    startangle=140, 
    colors=['#ff9999', '#66b3ff', '#99ff99']
)
plt.title('Percentage Breakdown of Student Age Groups')
plt.axis('equal') 
plt.savefig('Percentage Breakdown of Student Age Groups')
plt.show()
```
